```python
import sys 
sys.path.append("../src")
from generate_data import X_filled
import matplotlib.pyplot as plt

col = "Attribute 1"
data = X_filled[col]

mean = data.mean()
std = data.std()
ucl = mean + 3 * std
lcl = mean - 3 * std

print(f"Mean: {mean:.2f}")
print(f"UCL: {ucl:.2f}")
print(f"LCL: {lcl:.2f}")
```

```python
plt.figure(figsize=(12, 5))
plt.plot(data.values, marker="o", markersize=2, linestyle="-", linewidth=0.5)
plt.axhline(mean, color="green", label="Mean")
plt.axhline(ucl, color="red", linestyle="--", label="UCL")
plt.axhline(lcl, color="red", linestyle="--", label="LCL")
plt.title(f"Control Chart — {col}")
plt.xlabel("Observation (row order)")
plt.ylabel(col)
plt.legend()
plt.show()
```

```python
out_of_control = data[(data > ucl) | (data < lcl)]
print(f"{len(out_of_control)} out-of-control points out of {len(data)} total")
print(f"That's {len(out_of_control)/len(data)*100:.2f}% of readings")
```

```python
from generate_data import X_filled, Y
flagged_rows = out_of_control.index
failure_rate_flagged = Y.loc[flagged_rows].eq(1).mean() * 100
failure_rate_overall = Y.eq(1).mean() * 100

print(f"Failure rate among flagged points: {failure_rate_flagged:.1f}%")
print(f"Overall failure rate: {failure_rate_overall:.1f}%")
```
